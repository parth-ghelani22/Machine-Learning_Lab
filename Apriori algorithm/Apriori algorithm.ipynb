Practical 5: Apriori Algorithm for Association Rule Mining
Task: Find association rules between products using the Apriori algorithm in Python (mlxtend library).

```python
import pandas as pd

# Load the dataset
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_csv("/content/drive/MyDrive/MCA/Sem 3/ML/Practical/Practical-5/dataset for  finding association rules.csv")

# Display the first 5 rows of the DataFrame
display(df.head())
```

```python
# Convert the DataFrame into a list of lists, where each inner list represents a transaction
transactions = []
for i in range(0, df.shape[0]):
    transactions.append([str(df.values[i,j]) for j in range(0, df.shape[1]) if str(df.values[i,j]) != 'nan'])

# Display the first 5 transactions to verify the preprocessing
print("First 5 transactions:")
for i in range(5):
    print(transactions[i])

print(f"\nTotal number of transactions: {len(transactions)}")
```

```python
# Install mlxtend if not already installed
!pip install mlxtend

import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# Convert the list of transactions into a one-hot encoded DataFrame
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df_encoded = pd.DataFrame(te_ary, columns=te.columns_)

# Display the first 5 rows of the one-hot encoded DataFrame
print("\nFirst 5 rows of one-hot encoded DataFrame:")
display(df_encoded.head())
```

```python
# Apply the Apriori algorithm to find frequent itemsets
frequent_itemsets = apriori(df_encoded, min_support=0.01, use_colnames=True)

# Generate association rules from the frequent itemsets
association_rules_df = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.2)

# Display the frequent itemsets and association rules
print("\nFrequent Itemsets (first 5 rows):")
display(frequent_itemsets.head())

print("\nAssociation Rules (first 10 rows, sorted by confidence):")
display(association_rules_df.sort_values(by='confidence', ascending=False).head(10))
```
